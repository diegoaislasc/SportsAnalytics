```python
# Define the league and season parameters
# Note: Ensure 'EPL' and '23/24' match the expected format for ScraperFC's Sofascore module
LEAGUE = 'EPL'
SEASON = '21/22'

try:
    # 1. Retrieve match dictionaries for the entire season
    # This returns a list of dictionaries, where each dictionary represents a match
    match_dicts = ss.get_match_dicts(year=SEASON, league=LEAGUE)
    
    print(f"Retrieved {len(match_dicts)} match records.")

    # 2. Extract match IDs
    # Iterate through the list and extract the 'id' from each match dictionary
    match_ids = [match['id'] for match in match_dicts if 'id' in match]

    print(f"Extracted {len(match_ids)} match IDs.")
    
    # Optional: Display the first 5 IDs to verify
    print("Sample Match IDs:", match_ids[:5])
    
    # Optional: Check if the specific ID from your example exists in the list
    target_id = 9576454
    if target_id in match_ids:
        print(f"Target match ID {target_id} found in the dataset.")
    else:
        print(f"Target match ID {target_id} NOT found. (Check season/league parameters)")

except Exception as e:
    print(f"An error occurred: {e}")
```

```python

import ScraperFC as sfc
import inspect

def list_sofascore_methods():
    try:
        ss = sfc.Sofascore()
        methods = [method_name for method_name in dir(ss) if callable(getattr(ss, method_name)) and not method_name.startswith("_")]
        print("Available methods in ScraperFC.Sofascore:")
        for method in methods:
            print(f"- {method}")
            
            # optional: print signature
            try:
                sig = inspect.signature(getattr(ss, method))
                print(f"  Signature: {method}{sig}")
            except ValueError:
                pass
            print(f"  Doc: {getattr(ss, method).__doc__}")
            print()

    except Exception as e:
        print(f"Error inspecting ScraperFC.Sofascore: {e}")

if __name__ == "__main__":
    list_sofascore_methods()
```
